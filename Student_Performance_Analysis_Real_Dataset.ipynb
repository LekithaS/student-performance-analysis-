```python

import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Maths": [85, 72, 90, 65, 78, 92, 56, 81],
    "Science": [88, 75, 94, 70, 80, 89, 60, 84],
    "English": [78, 80, 85, 68, 75, 91, 62, 79]
}

df = pd.DataFrame(data)

df
```

```python

# 1. Display the first five rows
print("First five students:")
print(df.head())

# 2. Calculate average marks for each subject
print("\nAverage marks:")
print(df[["Maths", "Science", "English"]].mean())

# 3. Find each student's average
df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

print("\nStudent averages:")
print(df[["Student", "Average"]])

# 4. Find the student with the highest average
top_student = df.loc[df["Average"].idxmax()]

print("\nHighest average:")
print(top_student[["Student", "Average"]])
```

```python

import matplotlib.pyplot as plt

# Calculate average marks for each subject
subject_averages = df[["Maths", "Science", "English"]].mean()

# Create a bar chart
plt.figure(figsize=(8, 5))
plt.bar(subject_averages.index, subject_averages.values)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.show()
```

```python

# Compare marks scored by each student

df.plot(
    x="Student",
    y=["Maths", "Science", "English"],
    kind="bar",
    figsize=(10, 5)
)

plt.title("Subject-wise Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

## Key Findings

1. Science has the highest average marks among the three subjects in this sample dataset.
2. Student F has the highest average score across the three subjects.
3. Student G has the lowest average score across the three subjects.
4. Student performance varies across subjects, which can be observed in the comparison chart.

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/student-por.csv', sep=';')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

# 1. Check the number of rows and columns
print("Dataset shape:", df.shape)

# 2. Check all column names
print("\nColumn names:")
print(df.columns.tolist())

# 3. Check data types and missing values
print("\nDataset information:")
df.info()

# 4. Check missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# 5. Get statistical summaries
print("\nStatistical summary:")
display(df.describe())
```

```python

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check missing values
print("Total missing values:", df.isnull().sum().sum())

# Create a cleaned copy of the dataset
df_clean = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

# Check grade columns
print("\nGrade summary:")
display(df_clean[['G1', 'G2', 'G3']].describe())
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df_clean['G3'], bins=11, edgecolor='black')

plt.title('Distribution of Students’ Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')
plt.xticks(range(0, 21, 2))
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python

# Calculate overall grade statistics
average_grade = df_clean['G3'].mean()
highest_grade = df_clean['G3'].max()
lowest_grade = df_clean['G3'].min()
median_grade = df_clean['G3'].median()

print(f"Average final grade: {average_grade:.2f}")
print(f"Highest final grade: {highest_grade}")
print(f"Lowest final grade: {lowest_grade}")
print(f"Median final grade: {median_grade:.2f}")
```

```python

plt.figure(figsize=(8, 5))

plt.scatter(df_clean['G1'], df_clean['G3'],
            alpha=0.6, label='G1 vs G3')

plt.scatter(df_clean['G2'], df_clean['G3'],
            alpha=0.6, label='G2 vs G3')

plt.title('Relationship Between Earlier Grades and Final Grades')
plt.xlabel('Earlier Grades (G1 and G2)')
plt.ylabel('Final Grade (G3)')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python

# Calculate average final grade by sex
average_by_sex = df_clean.groupby('sex')['G3'].mean()

print("Average final grade by sex:")
display(average_by_sex)

# Visualize the comparison
average_by_sex.plot(
    kind='bar',
    figsize=(7, 5),
    edgecolor='black')

plt.title('Average Final Grade by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Final Grade')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python

# Average final grade by weekly study-time category
study_analysis = df_clean.groupby('studytime')['G3'].mean()

print("Average final grade by study-time category:")
display(study_analysis.round(2))

# Visualize the results
study_analysis.plot(
    kind='bar',
    figsize=(8, 5),
    edgecolor='black'
)

plt.title('Average Final Grade by Study Time')
plt.xlabel('Study Time Category (1 = Lowest, 4 = Highest)')
plt.ylabel('Average Final Grade')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python

# Calculate correlation between earlier and final grades
grade_correlation = df_clean[['G1', 'G2', 'G3']].corr()

print("Correlation between grades:")
display(grade_correlation.round(2))

# Visualize the correlation matrix
plt.figure(figsize=(6, 5))

plt.imshow(grade_correlation, vmin=-1, vmax=1, aspect='auto')
plt.colorbar(label='Correlation')
plt.xticks(range(3), grade_correlation.columns)
plt.yticks(range(3), grade_correlation.index)
plt.title('Correlation Between Student Grades')

# Display correlation values on the chart
for i in range(3):
    for j in range(3):
        plt.text(j, i, f'{grade_correlation.iloc[i, j]:.2f}',
                 ha='center', va='center')

plt.show()
```
