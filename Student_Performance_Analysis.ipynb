```python

import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Maths": [85, 72, 90, 65, 78, 92, 56, 81],
    "Science": [88, 75, 94, 70, 80, 89, 60, 84],
    "English": [78, 80, 85, 68, 75, 91, 62, 79]
}

df = pd.DataFrame(data)

df
```

```python

# 1. Display the first five rows
print("First five students:")
print(df.head())

# 2. Calculate average marks for each subject
print("\nAverage marks:")
print(df[["Maths", "Science", "English"]].mean())

# 3. Find each student's average
df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

print("\nStudent averages:")
print(df[["Student", "Average"]])

# 4. Find the student with the highest average
top_student = df.loc[df["Average"].idxmax()]

print("\nHighest average:")
print(top_student[["Student", "Average"]])
```

```python

import matplotlib.pyplot as plt

# Calculate average marks for each subject
subject_averages = df[["Maths", "Science", "English"]].mean()

# Create a bar chart
plt.figure(figsize=(8, 5))
plt.bar(subject_averages.index, subject_averages.values)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.show()
```

```python

# Compare marks scored by each student

df.plot(
    x="Student",
    y=["Maths", "Science", "English"],
    kind="bar",
    figsize=(10, 5)
)

plt.title("Subject-wise Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

## Key Findings

1. Science has the highest average marks among the three subjects in this sample dataset.
2. Student F has the highest average score across the three subjects.
3. Student G has the lowest average score across the three subjects.
4. Student performance varies across subjects, which can be observed in the comparison chart.
